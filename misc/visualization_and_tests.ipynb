# Klassische Portfolio-Optimierung

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, laplace

plt.style.use('seaborn-v0_8-whitegrid')

# -------------------------
# Effiziente Grenze + Portfolios
# -------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Simulierte effiziente Grenze
sigma = np.linspace(0.1, 0.3, 100)
mu = 0.15 + 0.4 * sigma**2  # Quadratische Approximation

ax1.plot(sigma, mu, color='#1f77b4', lw=2.5, label='Effiziente Grenze')
ax1.scatter([0.12, 0.18, 0.25], [0.16, 0.20, 0.23], 
           color='#ff7f0e', s=80, zorder=10,
           label='Mögliche Portfolios')

ax1.set_xlabel('Risiko (σ)', fontsize=12)
ax1.set_ylabel('Erwartete Rendite (μ)', fontsize=12)
ax1.set_title('Mean-Variance-Optimierung', pad=20)
ax1.legend(frameon=True, shadow=True)

# -------------------------
# Verteilungsvergleich
# -------------------------
x = np.linspace(-4, 4, 1000)
ax2.plot(x, norm.pdf(x), color='#9467bd', lw=2.5, 
        label='Normalverteilung (Theorie)')
ax2.plot(x, laplace.pdf(x, loc=0, scale=0.7), color='#d62728', lw=2.5,
        label='Fat-Tails (Praxis)')

# Fat-Tail Annotation
ax2.annotate('Höhere Extremrisiken', 
            xy=(2.8, 0.02), xytext=(2, 0.15),
            arrowprops=dict(arrowstyle="->", color='#2ca02c'),
            color='#2ca02c', fontsize=10)

ax2.set_xlabel('Standardisierte Renditeabweichungen', fontsize=12)
ax2.set_ylabel('Dichte', fontsize=12)
ax2.set_title('Verteilungsannahmen', pad=20)
ax2.legend()

plt.tight_layout()
plt.savefig('portfolio_analysis.pdf', bbox_inches='tight')  # Vektorgrafik
plt.show()
```
